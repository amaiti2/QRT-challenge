```python
RET_COLS = [f"RET_{i}" for i in range(1, 21)]
VOL_COLS = [f"SIGNED_VOLUME_{i}" for i in range(1, 21)]
WINDOWS = [2, 3, 5, 10, 20]


def safe_ratio(numerator, denominator):
    """Return NaN when a ratio has an invalid or near-zero denominator."""
    a, b = np.broadcast_arrays(np.asarray(numerator, dtype=float),np.asarray(denominator, dtype=float),)

    result = np.full(a.shape, np.nan)

    valid = (np.isfinite(a) & np.isfinite(b) & (np.abs(b) > 1e-12))

    np.divide(a, b, out=result, where=valid)
    return result


def row_correlation(a, b):
    """Pairwise-complete Pearson correlation for each row."""
    valid = np.isfinite(a) & np.isfinite(b)
    count = valid.sum(axis=1, keepdims=True)

    mean_a = (np.where(valid, a, 0.0).sum(axis=1, keepdims=True)/ np.maximum(count, 1))
    mean_b = (np.where(valid, b, 0.0).sum(axis=1, keepdims=True)/ np.maximum(count, 1))

    centered_a = np.where(valid, a - mean_a, 0.0)
    centered_b = np.where(valid, b - mean_b, 0.0)

    numerator = (centered_a * centered_b).sum(axis=1)
    denominator = np.sqrt((centered_a**2).sum(axis=1)* (centered_b**2).sum(axis=1))

    return safe_ratio(numerator, denominator)


def row_trend(recent_first):
    """OLS slope after reversing the sequence to oldest -> newest."""
    values = np.asarray(recent_first, dtype=float)[:, ::-1]

    times = np.broadcast_to(np.arange(values.shape[1], dtype=float),values.shape,)

    valid = np.isfinite(values)
    count = valid.sum(axis=1, keepdims=True)

    mean_time = (np.where(valid, times, 0.0).sum(axis=1, keepdims=True)/ np.maximum(count, 1))
    mean_value = (np.where(valid, values, 0.0).sum(axis=1, keepdims=True)/ np.maximum(count, 1))

    centered_time = np.where(valid, times - mean_time, 0.0)
    centered_value = np.where(valid, values - mean_value, 0.0)

    numerator = (centered_time * centered_value).sum(axis=1)
    denominator = (centered_time**2).sum(axis=1)

    return safe_ratio(numerator, denominator)


def make_history_features(df):
    z = df.copy()

    raw_numeric = RET_COLS + VOL_COLS + ["MEDIAN_DAILY_TURNOVER"]
    z[raw_numeric] = (z[raw_numeric].apply(pd.to_numeric, errors="raise")
        .replace([np.inf, -np.inf], np.nan))

    features = {}

    for k in WINDOWS:
        r = z[RET_COLS[:k]]
        v = z[VOL_COLS[:k]]

        mean_r = r.mean(axis=1)
        std_r = r.std(axis=1, ddof=1)

        features[f"ret_mean_{k}"] = mean_r
        features[f"ret_std_{k}"] = std_r
        features[f"ret_sum_{k}"] = r.sum(axis=1, min_count=1)
        features[f"ret_min_{k}"] = r.min(axis=1)
        features[f"ret_max_{k}"] = r.max(axis=1)
        features[f"ret_abs_mean_{k}"] = r.abs().mean(axis=1)

        # Missing returns are not counted as negative returns.
        positive = r.gt(0).where(r.notna())
        features[f"ret_positive_rate_{k}"] = positive.mean(axis=1).astype("float32")

        features[f"downside_rms_{k}"] = np.sqrt(
            r.clip(upper=0).pow(2).mean(axis=1)
        )

        features[f"ret_tstat_{k}"] = safe_ratio(mean_r,std_r / np.sqrt(r.count(axis=1)),)

        features[f"ret_trend_{k}"] = row_trend(r.to_numpy())

        features[f"volume_mean_{k}"] = v.mean(axis=1)
        features[f"volume_std_{k}"] = v.std(axis=1, ddof=1)
        features[f"volume_abs_mean_{k}"] = v.abs().mean(axis=1)
        features[f"volume_trend_{k}"] = row_trend(v.to_numpy())

        features[f"ret_volume_corr_{k}"] = row_correlation(r.to_numpy(),v.to_numpy(),)

        # Use arrays: r and v have different column names.
        products = pd.DataFrame(r.to_numpy() * v.to_numpy(),index=z.index,)
        features[f"ret_volume_product_{k}"] = products.mean(axis=1)

    z = pd.concat([z, pd.DataFrame(features, index=z.index)],axis=1,)

    z["short_long_return"] = z["ret_mean_3"] - z["ret_mean_20"]

    # Scale-normalized last return, not a centered z-score.
    z["last_return_scaled"] = safe_ratio(z["RET_1"], z["ret_std_20"])

    # A genuinely centered volume surprise.
    z["last_volume_surprise"] = safe_ratio(z["SIGNED_VOLUME_1"] - z["volume_mean_20"],
    z["volume_std_20"],)

    r = z[RET_COLS].to_numpy()
    pair_valid = np.isfinite(r[:, :-1]) & np.isfinite(r[:, 1:])

    sign_changes = (np.sign(r[:, :-1]) != np.sign(r[:, 1:])) & pair_valid

    z["sign_change_rate"] = safe_ratio(sign_changes.sum(axis=1),pair_valid.sum(axis=1),)

    turnover = z["MEDIAN_DAILY_TURNOVER"]
    z["turnover_x_volatility"] = turnover * z["ret_std_20"]
    z["turnover_x_momentum"] = turnover * z["ret_mean_5"]
    z["turnover_x_abs_volume"] = (turnover * z["volume_abs_mean_20"])

    return z
```

```python
# Goal: describe each allocation relative to its same-date peers.
# Mathematics: date means, cross-sectional z-scores, and ranks.

CONTEXT_COLUMNS = ["ret_mean_3","ret_mean_5","ret_mean_20","ret_std_20","ret_positive_rate_20",
    "volume_mean_20","ret_volume_corr_20","short_long_return",]


def add_cross_sectional_features(df):
    z = df.copy()
    additions = {}

    for col in CONTEXT_COLUMNS:
        date_group = z.groupby("TS", observed=True)[col]

        date_mean = date_group.transform("mean").astype("float32")
        date_std = date_group.transform("std").astype("float32")

        additions[f"date_mean_{col}"] = date_mean
        additions[f"date_z_{col}"] = safe_ratio(z[col] - date_mean,date_std,)
        additions[f"date_rank_{col}"] = date_group.rank(pct=True)

        peer_group = z.groupby(["TS", "GROUP"],observed=True,dropna=False,)[col]

        additions[f"group_z_{col}"] = safe_ratio(z[col] - peer_group.transform("mean"),
        peer_group.transform("std"),)
        additions[f"group_rank_{col}"] = peer_group.rank(pct=True)

    return pd.concat([z, pd.DataFrame(additions, index=z.index)],axis=1,)


train_features = add_cross_sectional_features(
    make_history_features(X_raw)
)
test_features = add_cross_sectional_features(
    make_history_features(X_test_raw)
)

categorical_columns = ["ALLOCATION", "GROUP"]

# Explicitly exclude identifiers and any target-like columns.
excluded = {"ROW_ID", "TS", "TARGET", "Y"}
feature_columns = [
    col for col in train_features.columns
    if col not in excluded
]
numeric_columns = [
    col for col in feature_columns
    if col not in categorical_columns
]

X = train_features[feature_columns].copy()
X_test = test_features[feature_columns].copy()

for col in categorical_columns:
    X[col] = X[col].fillna("__MISSING__").astype(str)
    X_test[col] = X_test[col].fillna("__MISSING__").astype(str)
```

```python
def add_optimized_features_inplace(df):
    """
    Add optimized historical features directly to df.

    The function changes df in place to avoid making a full copy
    of the 527,073-row dataset.
    """

    # Columns are arranged from most recent to oldest:
    #
    #       RET_1, RET_2, ..., RET_20.

    ret_columns = [f"RET_{lag}" for lag in range(1, 21)]

    volume_columns = [f"SIGNED_VOLUME_{lag}" for lag in range(1, 21)]

    # Check that all required historical variables exist.

    missing_columns = [column for column in ret_columns + volume_columns if column not in df.columns]

    if missing_columns:
        raise ValueError(f"Missing historical columns: {missing_columns}")

    # Convert the historical return matrix to float32: R ∈ R^(N × 20).
    # Row i is the 20-day return history of observation i.

    R = (df[ret_columns].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=np.float32))

    # Construct the signed-volume matrix: V ∈ R^(N × 20).

    V = (df[volume_columns].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=np.float32))

    def fill_missing_with_row_mean(A):
        """
        Replace missing historical values with the mean of the
        available observations in the same row.

        For row i: mean_i=
               sum_j A_ij 1[A_ij is observed]
               --------------------------------.
               sum_j 1[A_ij is observed]

        An entirely missing row is filled with zero.
        """

        observed = np.isfinite(A)

        number_observed = observed.sum(axis=1,keepdims=True,)

        row_sum = np.where(observed,A,0.0,).sum(axis=1, keepdims=True)

        row_mean = np.divide(row_sum,np.maximum(number_observed, 1),)

        return np.where(observed,A,row_mean,).astype(np.float32)

    R_filled = fill_missing_with_row_mean(R)
    V_filled = fill_missing_with_row_mean(V)

    # --------------------------------------------------------
    # 1. EXPONENTIALLY WEIGHTED RETURN
    # --------------------------------------------------------
    # Recent observations receive larger weights: w_j = 2^(-(j-1)/5),j = 1,...,20.
    # The feature is: ret_EWMA_i    =
    #       sum_j w_j RET_(i,j)
    #       --------------------.
    #             sum_j w_j

    ages = np.arange(20, dtype=np.float32)

    ewma_weights = np.power(0.5,ages / 5.0,).astype(np.float32)

    ewma_weights /= ewma_weights.sum()

    df["opt_ret_ewma_5"] = (R_filled @ ewma_weights).astype(np.float32)

    df["opt_volume_ewma_5"] = (V_filled @ ewma_weights).astype(np.float32)

    # --------------------------------------------------------
    # 4. ONE-DAY REVERSAL
    # --------------------------------------------------------
    # Compare the latest return with the preceding four days:
    # reversal_i= RET_1-mean(RET_2,...,RET_5).
    # Large positive or negative values indicate a recent shock.

    df["opt_reversal_1_5"] = (R_filled[:, 0]- R_filled[:, 1:5].mean(axis=1)).astype(np.float32)

    # --------------------------------------------------------
    # 5. TOTAL AND DOWNSIDE VOLATILITY
    # --------------------------------------------------------
    # Total volatility: sigma_i=sqrt[(1/20) sum_j (RET_(i,j) - mean_i)^2].
    # Downside volatility: sigma_i^-=sqrt[(1/20) sum_j min(RET_(i,j),0)^2].

    return_volatility = R_filled.std(axis=1)

    downside_volatility = np.sqrt(np.mean(np.minimum(R_filled, 0.0) ** 2,axis=1,))

    df["opt_downside_volatility_20"] = (downside_volatility.astype(np.float32))

    # --------------------------------------------------------
    # 6. RISK-ADJUSTED MOMENTUM
    # --------------------------------------------------------
    #
    #       risk_adjusted_momentum_i
    #       =
    #       momentum_i / (sigma_i + epsilon).
    #
    # Clipping prevents extremely small sigma_i from producing
    # unstable ratios.
    recent_return = R_filled[:, :3].mean(axis=1)
    older_return = R_filled[:, 3:].mean(axis=1)

    momentum = recent_return - older_return

    risk_adjusted_momentum = momentum / (return_volatility + 1e-8)

    df["opt_risk_adjusted_momentum"] = np.clip(risk_adjusted_momentum,-10.0,10.0,).astype(np.float32)

    # --------------------------------------------------------
    # 8. LAG-ONE RETURN AUTOCORRELATION
    # --------------------------------------------------------
    # This measures whether adjacent historical returns tend to
    # have the same or opposite direction:
    # rho_i =Corr((RET_1,...,RET_19),(RET_2,...,RET_20)).
    # rho_i > 0 suggests persistence.
    # rho_i < 0 suggests mean reversion.

    return_now = R_filled[:, :-1]
    return_lagged = R_filled[:, 1:]

    return_now_centered = (return_now - return_now.mean(axis=1, keepdims=True))

    return_lagged_centered = (return_lagged- return_lagged.mean(axis=1, keepdims=True))

    autocorrelation_numerator = np.sum(return_now_centered* return_lagged_centered,axis=1,)

    autocorrelation_denominator = np.sqrt(np.sum(return_now_centered**2, axis=1)
        * np.sum(return_lagged_centered**2, axis=1))

    autocorrelation = np.divide(autocorrelation_numerator,autocorrelation_denominator,
        out=np.zeros_like(autocorrelation_numerator,dtype=np.float32,),
        where=autocorrelation_denominator > 1e-12,)

    df["opt_ret_autocorrelation_1"] = (autocorrelation.astype(np.float32))

    # --------------------------------------------------------
    # 9. HISTORICAL MAXIMUM DRAWDOWN
    # --------------------------------------------------------
    # Approximate the cumulative historical allocation return:
    # C_(i,t) = sum_(s<=t) RET_(i,s).
    # Running drawdown:
    # D_(i,t) = C_(i,t) - max_(s<=t) C_(i,s).
    # Maximum drawdown is:
    # min_t D_(i,t) <= 0.

    R_chronological = R_filled[:, ::-1]
    V_chronological = V_filled[:, ::-1]
    cumulative_return = np.cumsum(R_chronological, axis=1,)

    running_maximum = np.maximum.accumulate(cumulative_return,axis=1,)

    drawdown = (cumulative_return - running_maximum)

    df["opt_max_drawdown_20"] = drawdown.min(axis=1).astype(np.float32)
    
    # --------------------------------------------------------
    # 11. RETURN-VOLUME ALIGNMENT
    # --------------------------------------------------------
    # alignment_i= (1/20) sum_j RET_(i,j) sign(VOLUME_(i,j)).
    # A positive value means the return direction generally
    # agrees with the signed-volume direction.

    df["opt_return_volume_alignment"] = np.mean( R_filled * np.sign(V_filled), axis=1,).astype(np.float32)

    # --------------------------------------------------------
    # 12. TURNOVER FEATURES
    # --------------------------------------------------------

    if "MEDIAN_DAILY_TURNOVER" in df.columns:

        turnover = pd.to_numeric(df["MEDIAN_DAILY_TURNOVER"], errors="coerce",
        ).to_numpy(dtype=np.float32)

        # Stabilized turnover:
        # log_turnover_i = log(1 + turnover_i).

        df["opt_log_turnover"] = np.log1p(np.maximum(turnover, 0.0)).astype(np.float32)

        # Interaction:
        # turnover_momentum_i=turnover_i |momentum_i|.
        # This measures whether a strong signal requires
        # substantial portfolio rebalancing.

        df["opt_turnover_x_abs_momentum"] = (turnover * np.abs(momentum)
        ).astype(np.float32)


# Add the same features to train and test.

add_optimized_features_inplace(X)
add_optimized_features_inplace(X_test)

# Record the newly engineered columns.

optimized_feature_columns = [column for column in X.columns if column.startswith("opt_")]


print("Number of optimized features:", len(optimized_feature_columns),)

print(optimized_feature_columns)
```

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler


def diagnose_hidden_date_order(X_train, X_test, train_dates, test_dates,target,):
    # --------------------------------------------------------
    # 1. COMBINE TRAIN AND TEST HISTORIES
    # --------------------------------------------------------

    return_columns = [f"RET_{lag}" for lag in range(1, 21)]

    train_data = X_train[["ALLOCATION"] + return_columns].reset_index(drop=True).copy()

    train_data["TS"] = np.asarray(train_dates)

    train_data["source"] = "train"

    train_data["source_row"] = np.arange(len(train_data))


    test_data = X_test[["ALLOCATION"] + return_columns].reset_index(drop=True).copy()

    test_data["TS"] = np.asarray(test_dates)

    test_data["source"] = "test"

    test_data["source_row"] = np.arange(len(test_data))


    combined = pd.concat([train_data, test_data],ignore_index=True,)

    combined["ALLOCATION"] = (combined["ALLOCATION"].astype(str))

    combined[return_columns] = (combined[return_columns].apply(pd.to_numeric, errors="coerce"))

    # --------------------------------------------------------
    # 2. CALCULATE THE MEAN RETURN HISTORY FOR EACH DATE
    # --------------------------------------------------------
    # mean_RET_(t,j) = average over allocations of RET_(S,t,j).

    date_mean_returns = (combined.groupby("TS")[return_columns].mean())

    date_labels = (date_mean_returns.index.to_numpy())

    # Outgoing history of source date:
    # RET_1,...,RET_19.

    outgoing = date_mean_returns[[f"RET_{lag}"for lag in range(1, 20)]].to_numpy()

    # Incoming history of possible successor:
    # RET_2,...,RET_20.

    incoming = date_mean_returns[[f"RET_{lag}"for lag in range(2, 21)]].to_numpy()

    print("date_mean_returns")
    print("Type:", type(date_mean_returns))
    print("Shape:", date_mean_returns.shape)
    print("Number of distinct dates:", len(date_labels))
    print("Columns:", date_mean_returns.columns.tolist())

    display(date_mean_returns.head())
    
    print("\noutgoing")
    print("Type:", type(outgoing))
    print("Shape:", outgoing.shape)

    print("\nincoming")
    print("Type:", type(incoming))
    print("Shape:", incoming.shape)
    
    outgoing_dataframe = pd.DataFrame(
    outgoing,
    index=date_labels,
    columns=[f"OUT_RET_{lag}" for lag in range(1, 20)],)

    incoming_dataframe = pd.DataFrame(incoming, index=date_labels,columns=[
        f"IN_RET_{lag}" for lag in range(2, 21)],)

    outgoing_dataframe.index.name = "TS"
    incoming_dataframe.index.name = "TS"

    print("\nOutgoing date signatures:")
    display(outgoing_dataframe.head())

    print("\nIncoming date signatures:")
    display(incoming_dataframe.head())
    
    # Replace missing date averages with zero.
    outgoing = np.nan_to_num(outgoing)
    incoming = np.nan_to_num(incoming)

    # --------------------------------------------------------
    # 3. STANDARDIZE THE SIGNATURES
    # --------------------------------------------------------

    scaler = StandardScaler()

    scaler.fit(np.vstack([outgoing, incoming]))

    outgoing_scaled = scaler.transform(outgoing)

    incoming_scaled = scaler.transform(incoming)

    print("outgoing_scaled shape:", outgoing_scaled.shape)
    print("incoming_scaled shape:", incoming_scaled.shape)

    outgoing_scaled_dataframe = pd.DataFrame(outgoing_scaled, index=date_labels,
    columns=[f"SCALED_OUT_{lag}" for lag in range(1, 20)],)

    incoming_scaled_dataframe = pd.DataFrame(incoming_scaled,index=date_labels,
    columns=[f"SCALED_IN_{lag}"for lag in range(2, 21)],)

    outgoing_scaled_dataframe.index.name = "TS"
    incoming_scaled_dataframe.index.name = "TS"

    print("\nStandardized outgoing signatures:")
    display(outgoing_scaled_dataframe.head())

    print("\nStandardized incoming signatures:")
    display(incoming_scaled_dataframe.head())
    # --------------------------------------------------------
    # 4. FIND THE NEAREST POSSIBLE SUCCESSOR DATE
    # --------------------------------------------------------

    number_of_dates = len(date_labels)

    number_of_neighbors = min(5,number_of_dates,)

    neighbor_model = NearestNeighbors(n_neighbors=number_of_neighbors)

    neighbor_model.fit(incoming_scaled)

    distances, positions = (neighbor_model.kneighbors(outgoing_scaled))

    successor_positions = np.full(number_of_dates,-1,dtype=int,)

    successor_distances = np.full(number_of_dates,np.nan,)

    second_distances = np.full(number_of_dates,np.nan,)


    for source_position in range(number_of_dates):

        # Exclude matching a date with itself.

        candidates = [(distance, position) for distance, position in zip(distances[source_position],
                positions[source_position],)
            if position != source_position]

        if len(candidates) >= 1:
            successor_distances[source_position] = candidates[0][0]

            successor_positions[source_position] = candidates[0][1]

        if len(candidates) >= 2:
            second_distances[source_position] = candidates[1][0]


    successor_dates = [(date_labels[position] if position >= 0 else None)
        for position in successor_positions]

    date_mapping = pd.DataFrame(
        {"TS": date_labels, "SuccessorTS": successor_dates, "Distance": successor_distances,

            # Smaller ratios indicate clearer matches.

         "DistanceRatio": (successor_distances / (second_distances + 1e-12)),})
    
    
    # --------------------------------------------------------
    # 5. LOOK UP RET_1 FOR THE SAME ALLOCATION ON THE
    #    INFERRED SUCCESSOR DATE
    # --------------------------------------------------------

    successor_date_map = (date_mapping.set_index("TS")["SuccessorTS"])

    distance_map = (date_mapping.set_index("TS")["Distance"])


    next_return_lookup = (combined.groupby(["TS", "ALLOCATION"])["RET_1"].mean())


    training_rows = combined[combined["source"] == "train"].copy()

    training_rows["SuccessorTS"] = (training_rows["TS"].map(successor_date_map))

    training_rows["MatchDistance"] = (training_rows["TS"].map(distance_map))


    lookup_index = pd.MultiIndex.from_arrays([training_rows["SuccessorTS"],training_rows["ALLOCATION"],
        ],names=["TS", "ALLOCATION"],)

    training_rows["RevealedReturn"] = (next_return_lookup.reindex(lookup_index).to_numpy())

    training_rows = training_rows.sort_values("source_row")

    # --------------------------------------------------------
    # 6. EVALUATE MATCH QUALITY
    # --------------------------------------------------------

    target_array = np.asarray(target).reshape(-1).astype(int)

    evaluation = []

    # Evaluate all matches and increasingly strict distance
    # subsets.

    for distance_quantile in [0.10,0.25,0.50,1.00,]:

        cutoff = date_mapping["Distance"].quantile(distance_quantile)

        selected = (training_rows["RevealedReturn"].notna()
            & (training_rows["MatchDistance"]<= cutoff))

        number_selected = int(selected.sum())

        if number_selected > 0:

            row_numbers = training_rows.loc[selected,"source_row",].astype(int).to_numpy()

            predicted_target = (training_rows.loc[selected,"RevealedReturn",].to_numpy()> 0).astype(int)

            accuracy = np.mean(predicted_target == target_array[row_numbers])

        else:
            accuracy = np.nan

        evaluation.append({"DistanceQuantile": distance_quantile, "RowsMatched": number_selected,
                "Coverage": number_selected / len(X_train), "Accuracy": accuracy,})

    return ( date_mapping, pd.DataFrame(evaluation), training_rows,)
```

```python
# Import NumPy for arrays and numerical calculations.
import numpy as np

# Import accuracy and log-loss evaluation functions.
from sklearn.metrics import accuracy_score, log_loss

# Import grouped cross-validation.
from sklearn.model_selection import GroupKFold

# Extract the date identifier t_i for every training observation.
groups = X_raw["TS"].to_numpy()

# Use five date-grouped validation folds, so K = 5.
N_SPLITS = 5

# Create the grouped cross-validation rule.
# GroupKFold ensures that one date cannot occur in both T_k and V_k.
cv = GroupKFold(n_splits=N_SPLITS)

# Generate all K pairs (T_k, V_k).
# X is used only to determine the number of observations.
# y is the binary target.
# groups supplies the date associated with each observation.
folds = list(cv.split(X=X,y=y,groups=groups,))

def select_rows(data, indices):
    """Select observations with positions contained in `indices`.
If data is a pandas DataFrame, use `.iloc`.If data is a NumPy array, use ordinary NumPy indexing.
This allows the same cross-validation function to work with:
1. tabular feature matrices;
2. MiniROCKET sequence arrays.
    """
# Pandas objects possess the attribute `.iloc`.
    if hasattr(data, "iloc"):
# Select DataFrame rows by integer position.
        return data.iloc[indices]
# Otherwise, assume that `data` is a NumPy array.
    return data[indices]

def evaluate_model(name, actual, probability, threshold=0.5):
    """Evaluate probability predictions.
    Parameters
    name:Label used when printing the results.
    actual:True binary outcomes Y_i.
    probability:Predicted probabilities p_hat_i.
    threshold:Classification threshold c.

    Mathematics
    Predicted class: Y_hat_i(c) = 1{p_hat_i >= c}.
    Accuracy: A(c) = (1/N) sum_i 1{Y_hat_i(c) = Y_i}.
    Log-loss: L = -(1/N) sum_i [Y_i log(p_hat_i)+ (1-Y_i)log(1-p_hat_i)].
    """

    # Prevent log(0), which is mathematically infinite.
    # The clipped value satisfies:
    # 10^(-7) <= p_hat_i <= 1 - 10^(-7).
    probability = np.clip(probability,1e-7,1 - 1e-7,)

    # Apply the binary decision rule:
    # Y_hat_i = 1{p_hat_i >= threshold}.
    prediction = (probability >= threshold).astype(int)

    # Calculate:
    # Accuracy = mean(1{Y_hat_i = Y_i}).
    accuracy = accuracy_score(actual,prediction,)

    # Calculate average binary cross-entropy.
    loss = log_loss(actual,probability,labels=[0, 1],)

    # Display the model name and its two validation metrics.
    print(f"{name}: " f"accuracy={accuracy:.6f}, " f"logloss={loss:.6f}")

    # Return both values so that they can be stored if necessary.
    return accuracy, loss


def cross_validate_model(model_name, model_builder, training_data, test_data,):
    """Fit and evaluate one model using date-grouped cross-validation.
    Mathematical output
    
    oof_probability[i]:
    p_hat_i^OOF = p_hat^(-k)(x_i),  i in V_k.

    test_probability[j]:
    p_hat_test(x_j)= (1/K) sum_{k=1}^K p_hat^(-k)(x_j).

    The OOF predictions can be used to compare models without
    evaluating a model on observations used to fit that model.
    """

    # Allocate a vector for the N out-of-fold probabilities.
    oof_probability = np.zeros(len(training_data))

    # Allocate a vector for the test probabilities.
    test_probability = np.zeros(len(test_data))

    # Iterate through the K date-grouped folds.
    for fold_number, (train_idx, valid_idx) in enumerate(folds,start=1,):

        # Construct a fresh, unfitted model for this fold.
        model = model_builder()

        # Select T_k, the training observations for fold k.
        X_train_fold = select_rows(training_data,train_idx,)

        # Select V_k, the validation observations for fold k.
        X_valid_fold = select_rows(training_data,valid_idx,)

        # Fit p_hat^(-k) using only observations in T_k.
        model.fit(X_train_fold,y[train_idx],)

        # Predict p_hat^(-k)(x_i) for every i in V_k.
        oof_probability[valid_idx] = model.predict_proba(X_valid_fold)[:, 1]

        # Add (1/K) p_hat^(-k)(x) to the test probability average.
        test_probability += (model.predict_proba(test_data)[:, 1]/ N_SPLITS)

        # Evaluate this fold using observations in V_k.
        evaluate_model(name=f"{model_name}, fold {fold_number}",actual=y[valid_idx],
            probability=oof_probability[valid_idx],)

    # Evaluate the combined OOF predictions for all N observations.
    evaluate_model(name=f"{model_name}, complete OOF",actual=y,probability=oof_probability,)

    # Return training OOF probabilities and averaged test probabilities.
    return oof_probability, test_probability


# Calculate the positive-class frequency:
#pi_hat = (1/N) sum_i Y_i.
positive_rate = y.mean()

# The majority-class baseline predicts the more frequent class.
# Its accuracy is:
# max(pi_hat, 1 - pi_hat).
majority_accuracy = max(positive_rate,1 - positive_rate,)

# Print the minimum baseline that a useful model should exceed.
print("Positive-class frequency:", positive_rate)
print("Majority-class accuracy:", majority_accuracy)
```

```python
# Import a transformer that applies different operations to
# numerical and categorical columns.
from sklearn.compose import ColumnTransformer

# Import median imputation for missing numerical observations.
from sklearn.impute import SimpleImputer

# Import regularized logistic regression.
from sklearn.linear_model import LogisticRegression

# Import Pipeline so preprocessing and estimation are fitted together.
from sklearn.pipeline import Pipeline

# Import standardization and one-hot encoding.
from sklearn.preprocessing import OneHotEncoder, StandardScaler


def build_logistic_model():
    """Create a new logistic-regression pipeline.
    Because this function is called separately for each fold,
    the imputation medians and scaling parameters are estimated
    using T_k only.
    """
# Construct numerical preprocessing.
    numeric_processing = Pipeline([("impute",SimpleImputer(strategy="median"),),("scale",StandardScaler(),),])
# Replace a missing x_ij with the training-fold median of feature j.
# Transform x_ij into:z_ij = (x_ij - mu_j) / s_j.
        

# Apply numerical processing to numerical columns and one-hot encoding to categorical columns.
    preprocessing = ColumnTransformer([("numeric",numeric_processing,numeric_columns,),
("categorical",OneHotEncoder(handle_unknown="ignore"),categorical_columns,),])
# Numerical branch of phi(x).    
# Categorical branch of phi(x). handle_unknown="ignore" means that an unseen test
# category receives zero in all known-category indicators.    

# Define the regularized logistic classifier.
    classifier = LogisticRegression(C=0.1, solver= "saga", max_iter=20, random_state=42)
# C = 0.1 means relatively strong L2 regularization.
# SAGA works efficiently with large sparse feature matrices.
# Maximum number of optimization iterations.
# Fix the stochastic elements for reproducibility.

# Combine preprocessing and classification.
# During model.fit():x_i -> phi(x_i) = z_i -> p_hat_i.
# Estimate and apply phi.    
# Estimate beta_0 and beta.    
    return Pipeline([("preprocessing",preprocessing,),("classifier",classifier,),])


# Generate:
# logit_oof[i]  = out-of-fold logistic probability;
# logit_test[j] = average test logistic probability.
logit_oof, logit_test = cross_validate_model(
    model_name="Logistic regression",
    model_builder=build_logistic_model,
    training_data=X,
    test_data=X_test,)
```

```python


# as we can see in the previous codeblock of Logistic regression, we have a problem of stabilizing
# the coefficients for 30 iteraions(same with 300) although it is taking quite a bit of time and memory
# So let us use the SGDClassifier instead, where at each optimization step, we use a smaller dataset,
# hence far more usefull and cost effective in terms of memory and time
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier


# Use a compact collection of informative numerical features.
# This avoids one-hot encoding the high-cardinality ALLOCATION
# column and substantially reduces fitting time and memory use.

# These variables summarize:
#   - recent returns,
#   - return momentum,
#   - volatility,
#   - liquidity,
#   - turnover,
#   - date context,
#   - group context.

candidate_logistic_columns = [
    # Most recent raw observations
    "RET_1","SIGNED_VOLUME_1","MEDIAN_DAILY_TURNOVER",

    # Historical return level:
    # ret_mean_h = (1/h) sum_(j=1)^h RET_j
    "ret_mean_3","ret_mean_5","ret_mean_20",

    # Historical return volatility:
    # ret_std_h = sqrt[(1/h) sum_(j=1)^h(RET_j - ret_mean_h)^2]
    "ret_std_5","ret_std_20",

    # Fraction of positive historical returns:
    # ret_positive_rate_20 = (1/20) sum_(j=1)^20 1[RET_j > 0]
    "ret_positive_rate_20",

    # Liquidity level and variability
    "volume_mean_20","volume_std_20",

    # Interaction between returns and signed volume
    "ret_volume_corr_20",

    # Difference between short- and long-term momentum
    "short_long_return",

    # Standardized recent return and volume
    "last_return_scaled","last_volume_surprise",

    # Turnover interactions
    "turnover_x_volatility","turnover_x_momentum",

    # Cross-sectional date context
    "date_mean_ret_mean_5","date_z_ret_mean_5","date_rank_ret_mean_5",

    # Allocation-group context
    "group_z_ret_mean_5","group_rank_ret_mean_5",]


# Keep only names that were actually created in Section 2.
# Mathematically, this produces the selected feature index set:
# J = {j : feature j exists in X}

# Each observation is then represented by:
# x_i = (X_ij)_(j in J)

# Therefore:
# x_i is in R^d, where: d = len(logistic_columns).

logistic_columns = [column for column in candidate_logistic_columns if column in X.columns]
logistic_columns = list(
    dict.fromkeys(
        logistic_columns
        + optimized_feature_columns
    )
)


# Keep numerical columns only for the fast SGD logistic model.

logistic_columns = [
    column
    for column in logistic_columns
    if column in X.columns
    and pd.api.types.is_numeric_dtype(X[column])
]


X_logistic = X[logistic_columns]
X_test_logistic = X_test[logistic_columns]

print(
    "Optimized logistic feature count:",
    len(logistic_columns),
)
print("Number of logistic-regression features:", len(logistic_columns))


# Select the same columns, in the same order, from train and test.
# Dimensions:
# X_logistic      in R^(527073 × d)
# X_test_logistic in R^(31870 × d)


# Construct the preprocessing and logistic-regression pipeline.

fast_logistic_model = Pipeline(
    steps=[
        (
            "imputer",

            # Replace a missing value X_ij with the median of
            # feature j calculated from the training fold:X_ij <- median(X_1j, ..., X_nj)
            # The median is relatively resistant to extreme values.

            SimpleImputer(strategy="median"),
        ),
        (
            "scaler",

            # Standardize each feature:z_ij = (X_ij - mean_j) / std_j
            # This is important because SGD uses gradients.
            # Without scaling, large-valued features could create
            # much larger gradient updates than small-valued features.

            StandardScaler(),
        ),
        (
            "model",

            SGDClassifier(
                # loss="log_loss" makes this stochastic logistic
                # regression and enables predict_proba().
                loss="log_loss",

                # L2 regularization adds:
                # (alpha/2) ||beta||_2^2
                # to the objective function.
                penalty="l2",
                alpha=1e-4,

                # Make no more than 30 passes through the training
                # observations in each cross-validation fold.
                max_iter=300,

                # Stop when the improvement is sufficiently small.
                tol=1e-4,

                # Reserve part of the current training fold to
                # decide when optimization should stop.
                early_stopping=True,
                validation_fraction=0.20,
                n_iter_no_change=5,

                # Average coefficients across SGD updates:
                # beta_bar = (1/K) sum_(k=1)^K beta_k
                # This usually makes stochastic estimates more stable.
                average=True,

                random_state=42,
            ),
        ),
    ]
)


# Fit one model for every date-based cross-validation fold.
# For fold k:
#   1. Fit the model on training dates T_k.
#   2. Predict validation probabilities:
#          p_hat_i = P(y_i = 1 | x_i),   i in V_k
#   3. Convert probabilities into signs:
#          y_hat_i = 1[p_hat_i >= 0.5]
#   4. Compute validation accuracy:
#          Accuracy_k = (1 / |V_k|) sum_(i in V_k) 1[y_hat_i = y_i]
# The function also averages test probabilities across folds:
#p_hat_test= (1/K) sum_(k=1)^K p_hat_test^(k)

logit_oof, logit_test = cross_validate_model(
    model_name="Fast SGD logistic regression",
    model_builder=lambda: fast_logistic_model,
    training_data=X_logistic,
    test_data=X_test_logistic,)
```

```python
# Import a fast logistic classifier trained with gradient descent.
from sklearn.linear_model import SGDClassifier

%pip install aeon
# Import the MiniROCKET sequence classifier.
from aeon.classification.convolution_based import MiniRocketClassifier


def create_sequences(df):
    """Convert one DataFrame into an array with shape: observations x channels x timepoints.
    The resulting shape is: number of rows x 2 x 20.
    """

    # Extract return histories.
    # Reverse the columns so time goes from oldest to newest:
    # RET_20, RET_19, ..., RET_1.
    returns = (df[RET_COLS[::-1]].fillna(0.0).to_numpy(dtype=np.float32))

    # Extract signed-volume histories in the same chronological order:
    # SIGNED_VOLUME_20, ..., SIGNED_VOLUME_1.
    volumes = (df[VOL_COLS[::-1]].fillna(0.0).to_numpy(dtype=np.float32))

    # Stack the two matrices as two channels.
    # If there are N observations, the resulting shape is:(N, 2, 20).
    sequences = np.stack([returns, volumes],axis=1,)

    # Calculate mean_i,c for every observation and channel.
    channel_mean = sequences.mean(axis=2,keepdims=True,)

    # Calculate std_i,c for every observation and channel.
    channel_std = sequences.std(axis=2,keepdims=True,)

    # Prevent division by zero for constant sequences.
    channel_std = np.maximum(channel_std,1e-8,)

    # Return:A_tilde_i,c,j= (A_i,c,j - mean_i,c) / std_i,c.
    return (sequences - channel_mean) / channel_std


# Construct the N x 2 x 20 training sequence array.
X_sequence = create_sequences(X)

# Construct the N_test x 2 x 20 test sequence array.
X_test_sequence = create_sequences(X_test)


def build_rocket_model():
    """
    Construct MiniROCKET followed by a logistic classifier.
    """

    # Define the logistic classifier that acts on the
    # transformed MiniROCKET feature vectors z_i.
    logistic_classifier = SGDClassifier(

        # Minimize logistic/cross-entropy loss.
        loss="log_loss",

        # Add an L2 penalty:(alpha/2) ||theta||_2^2.
        penalty="l2",

        # Regularization strength.
        alpha=1e-3,

        # Maximum number of optimization passes.
        max_iter=200,

        # Fix randomness for reproducibility.
        random_state=42,
    )

    # Construct the sequence transformation and classifier.
    return MiniRocketClassifier(

        # Create approximately 50 convolution-derived features.
        #
        # Start small because N is large.
        n_kernels=500,

        # Use the logistic classifier defined above.
        estimator=logistic_classifier,

        # Use all available CPU cores.
        n_jobs=-1,

        # Make kernel construction reproducible.
        random_state=42,
    )


# Generate MiniROCKET OOF and test probabilities.
rocket_oof, rocket_test = cross_validate_model(
    model_name="MiniROCKET",
    model_builder=build_rocket_model,
    training_data=X_sequence,
    test_data=X_test_sequence,
)
```

```python
import numpy as np
from scipy.optimize import minimize
from scipy.special import expit, logit
from sklearn.metrics import log_loss, accuracy_score

y_array = np.asarray(y).reshape(-1).astype(int)

# Each column is one model's predicted probability.
oof_probs = np.column_stack([logit_oof, cat_oof, lgb_oof])
test_probs = np.column_stack([logit_test, cat_test, lgb_test])

# Blend on the log-odds scale.
oof_logits = logit(np.clip(oof_probs, 1e-6, 1 - 1e-6))
test_logits = logit(np.clip(test_probs, 1e-6, 1 - 1e-6))

result = minimize(
    lambda weights: log_loss(
        y_array, expit(oof_logits @ weights)
    ) + 0.001 * np.sum(weights**2),
    x0=[1/6, 1/2, 1/3],
    method="SLSQP",
    bounds=[(0, 1)] * 3,
    constraints={"type": "eq", "fun": lambda weights: weights.sum() - 1},
)

if not result.success:
    raise RuntimeError(result.message)

ensemble_weights = result.x
optimized_ensemble_oof = expit(oof_logits @ ensemble_weights)
optimized_ensemble_test = expit(test_logits @ ensemble_weights)
ensemble_threshold = 0.5

print("Weights [logistic, CatBoost, LightGBM]:", ensemble_weights)
print(
    "Ensemble OOF accuracy:",
    accuracy_score(y_array, optimized_ensemble_oof >= ensemble_threshold),
)
```

```python
from sklearn.metrics import roc_curve, accuracy_score
from sklearn.model_selection import GroupShuffleSplit
from scipy.special import expit
import numpy as np

def find_accuracy_weights_and_cutoff(row_indices):
    labels = y_array[row_indices]
    logits = oof_logits[row_indices]

    best_accuracy = -1
    best_weights = None
    best_cutoff = None

    # Nonnegative weights summing to 1, in steps of 0.05.
    for i in range(21):
        for j in range(21 - i):
            weights = np.array([i, j, 20 - i - j]) / 20
            scores = logits @ weights

            # Check every distinct score as a possible cutoff.
            fpr, tpr, cutoffs = roc_curve(
                labels, scores, drop_intermediate=False
            )

            positive_fraction = labels.mean()
            accuracies = (
                positive_fraction * tpr
                + (1 - positive_fraction) * (1 - fpr)
            )

            position = np.argmax(accuracies)

            if accuracies[position] > best_accuracy:
                best_accuracy = accuracies[position]
                best_weights = weights.copy()
                best_cutoff = cutoffs[position]

    return best_weights, best_cutoff, best_accuracy


# Choose weights/cutoff on one set of dates;
# check them on dates not used in this search.
tune_idx, check_idx = next(
    GroupShuffleSplit(
        n_splits=1, test_size=0.20, random_state=42
    ).split(
        oof_logits,
        y_array,
        groups=X_raw["TS"].to_numpy(),
    )
)

weights, score_cutoff, tuning_accuracy = (
    find_accuracy_weights_and_cutoff(tune_idx)
)

check_prediction = (
    oof_logits[check_idx] @ weights >= score_cutoff
).astype(int)

print("Weights [logistic, CatBoost, LightGBM]:", weights)
print("Ensemble probability threshold:", expit(score_cutoff))
print("Accuracy on tuning dates:", tuning_accuracy)
print(
    "Accuracy on check dates:",
    accuracy_score(y_array[check_idx], check_prediction),
)

# Use these choices for test predictions.
ensemble_weights = weights
ensemble_threshold = expit(score_cutoff)

optimized_ensemble_oof = expit(oof_logits @ ensemble_weights)
optimized_ensemble_test = expit(test_logits @ ensemble_weights)
```
